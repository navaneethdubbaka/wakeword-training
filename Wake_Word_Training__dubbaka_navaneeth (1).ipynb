```python
# =====================================================
# CONVERT + DUPLICATE CLIPS EVENLY
# =====================================================

import librosa, soundfile as sf, os, math

INPUT_DIR = "my_recordings_raw"
OUTPUT_DIR = f"generated_clips/{WAKEWORD}"

TARGET_TOTAL_CLIPS = 600   # 👈 CHANGE THIS NUMBER

os.makedirs(OUTPUT_DIR, exist_ok=True)

# Collect original wav files
files = [f for f in sorted(os.listdir(INPUT_DIR)) if f.endswith(".wav")]
num_original = len(files)

if num_original == 0:
    raise Exception("❌ No WAV files found in my_recordings_raw")

print("Original clips:", num_original)

# Calculate how many times each file should repeat
dup_per_file = math.ceil(TARGET_TOTAL_CLIPS / num_original)

print("Each clip will repeat:", dup_per_file, "times")

count = 0

for fname in files:
    audio, _ = librosa.load(os.path.join(INPUT_DIR, fname), sr=16000, mono=True)

    for _ in range(dup_per_file):
        out_path = os.path.join(OUTPUT_DIR, f"clip_{count:04d}.wav")

        sf.write(out_path, audio, 16000, subtype="PCM_16")

        count += 1

        if count >= TARGET_TOTAL_CLIPS:
            break

    if count >= TARGET_TOTAL_CLIPS:
        break

print("✅ Total generated clips:", count)
```

```python
# =====================================================
# COPY ALL GENERATED CLIPS INTO TRAIN / TEST
# =====================================================

import shutil, os

clips_dir = f"generated_clips/{WAKEWORD}"

pos_train = f"my_custom_model/{WAKEWORD}/positive_train"
pos_test = f"my_custom_model/{WAKEWORD}/positive_test"

# Clear old files first (IMPORTANT)
for folder in [pos_train, pos_test]:
    for f in os.listdir(folder):
        os.remove(os.path.join(folder, f))

# Copy ALL duplicated clips
count = 0

for f in os.listdir(clips_dir):
    if f.endswith(".wav"):
        shutil.copy(f"{clips_dir}/{f}", pos_train)
        shutil.copy(f"{clips_dir}/{f}", pos_test)
        count += 1

print("✅ Total copied to positive folders:", count)
```

```python
# =====================================================
# COPY POSITIVE + CREATE NEGATIVE SAMPLES
# =====================================================

import shutil, numpy as np, scipy.io.wavfile as wav

pos_train = f"my_custom_model/{WAKEWORD}/positive_train"
pos_test = f"my_custom_model/{WAKEWORD}/positive_test"

neg_train = f"my_custom_model/{WAKEWORD}/negative_train"
neg_test = f"my_custom_model/{WAKEWORD}/negative_test"

clips_dir = f"generated_clips/{WAKEWORD}"

# Copy positives
for f in os.listdir(clips_dir):
    shutil.copy(f"{clips_dir}/{f}", pos_train)
    shutil.copy(f"{clips_dir}/{f}", pos_test)

# Create silence negatives
sr = 16000
samples = np.zeros(sr*2, dtype=np.int16)

for i in range(100):
    wav.write(f"{neg_train}/silence_{i}.wav", sr, samples)
    wav.write(f"{neg_test}/silence_{i}.wav", sr, samples)

print("✅ Dataset ready")
```

```python
# =====================================================
# FIX TORCHAUDIO API COMPATIBILITY FOR OPENWAKEWORD
# =====================================================

import torchaudio
import soundfile as sf

# Patch missing info() function
if not hasattr(torchaudio, "info"):
    def fake_info(path):
        data, sr = sf.read(path)
        class Info:
            num_frames = len(data)
            sample_rate = sr
        return Info()
    torchaudio.info = fake_info

print("✅ torchaudio.info patched successfully")
```
